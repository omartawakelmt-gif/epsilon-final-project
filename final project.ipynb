# Online Shoppers Purchasing Intention Dataset

## 1. Dataset Overview

The **Online Shoppers Purchasing Intention Dataset** contains information about visitors' browsing behavior on an e-commerce website.

The main purpose of this dataset is to analyze customer behavior and predict whether a visitor will complete a purchase.

- **Total Rows:** 12,330
- **Total Columns:** 18
- **Input Features:** 17
- **Target Variable:** `Revenue`
- **Problem Type:** Binary Classification

Each row represents a different website browsing session.

## 2. Column Descriptions

| Column | Data Type | Description |
|---|---|---|
| `Administrative` | Integer | Number of administrative pages visited during the session, such as account management or login pages. |
| `Administrative_Duration` | Float | Total time spent on administrative pages, measured in seconds. |
| `Informational` | Integer | Number of informational pages visited, such as contact information, FAQs, or company details. |
| `Informational_Duration` | Float | Total time spent on informational pages, measured in seconds. |
| `ProductRelated` | Integer | Number of product-related pages visited during the session. |
| `ProductRelated_Duration` | Float | Total time spent browsing product-related pages, measured in seconds. |
| `BounceRates` | Float | Average bounce rate of the pages visited. It represents the proportion of visits that leave the website without further interaction. |
| `ExitRates` | Float | Average exit rate of the pages visited. It represents how frequently visitors leave the website from those pages. |
| `PageValues` | Float | Average monetary value associated with the pages visited before completing an e-commerce transaction. Higher values may indicate stronger purchasing intent. |
| `SpecialDay` | Float | A value between 0 and 1 representing how close the visit is to a special shopping date, such as Valentine's Day or Mother's Day. Higher values indicate greater proximity or relevance to the special day. |
| `Month` | Categorical | Month in which the browsing session occurred, such as Feb, Mar, May, or Nov. |
| `OperatingSystems` | Integer (Categorical) | Encoded identifier representing the visitor's operating system. The numbers represent categories rather than numerical quantities. |
| `Browser` | Integer (Categorical) | Encoded identifier representing the web browser used by the visitor. |
| `Region` | Integer (Categorical) | Encoded identifier representing the geographical region of the visitor. |
| `TrafficType` | Integer (Categorical) | Encoded identifier representing the source or type of traffic that brought the visitor to the website. |
| `VisitorType` | Categorical | Type of website visitor, such as `Returning_Visitor`, `New_Visitor`, or `Other`. |
| `Weekend` | Boolean | Indicates whether the session occurred on a weekend (`True`) or a weekday (`False`). |
| `Revenue` | Boolean | Target variable indicating whether the browsing session resulted in a purchase (`True`) or not (`False`). |

## 3. Feature Categories

The dataset features can be divided into the following groups:

### A. Visitor Browsing Behavior

These features describe how visitors interact with different types of website pages:

- `Administrative`
- `Administrative_Duration`
- `Informational`
- `Informational_Duration`
- `ProductRelated`
- `ProductRelated_Duration`

These features help measure visitor engagement and browsing activity.

### B. Website Engagement Metrics

These features describe how visitors interact with the website and how valuable their visits may be:

- `BounceRates`
- `ExitRates`
- `PageValues`

These metrics can help identify patterns associated with purchasing behavior.

### C. Time-Related Features

These features describe when a visitor accessed the website:

- `SpecialDay`
- `Month`
- `Weekend`

They can help identify seasonal shopping patterns and differences between weekday and weekend behavior.

### D. Technical and Visitor Information

These features describe the visitor's technical environment, geographical category, and visitor characteristics:

- `OperatingSystems`
- `Browser`
- `Region`
- `TrafficType`
- `VisitorType`

Although some of these features are represented by integers, they should be treated as categorical variables because their numerical values are category identifiers.

## 4. Target Variable: Revenue

The target variable for this dataset is **`Revenue`**.

It indicates whether a visitor completed a purchase during their browsing session.

| Value | Meaning |
|---|---|
| `True` | The visitor completed a purchase. |
| `False` | The visitor did not complete a purchase. |

The goal of the machine learning model is to predict `Revenue` based on the other 17 features.

## 5. Machine Learning Objective

The main objective is to build a **binary classification model** that predicts whether an online shopper will make a purchase.

This can help e-commerce businesses:

- Understand customer browsing behavior.
- Identify visitors with a high probability of purchasing.
- Improve website design and user experience.
- Develop more effective marketing strategies.
- Increase conversion rates.

**Target Variable:** `Revenue`

**Independent Variables:** All remaining 17 columns.

**Expected Output:** A prediction indicating whether a browsing session will result in a purchase (`True`) or no purchase (`False`).

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder

```

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore",drop="first")
scaler = StandardScaler()
robust = RobustScaler()

```

```python
robust_pip = Pipeline(steps=[("robust", robust)])
robust_pip
```

```python
scaler_pip = Pipeline(steps=[("scaler", scaler)])
scaler_pip
```

```python
ohe_pip = Pipeline(steps=[("ohe", ohe)])
ohe_pip
```

```python
nummeric_columns
```

```python
robust_columns = ["Administrative_Duration","Informational_Duration","ProductRelated","ProductRelated_Duration","BounceRates","PageValues"]
scaler_columns = ["Administrative","Informational","ExitRates","SpecialDay"]
```

```python
category_columns 
```

```python
from sklearn.compose import ColumnTransformer
preprocessing = ColumnTransformer(transformers=[("robust", robust_pip, robust_columns), 
                                              ("scaler", scaler_pip, scaler_columns),
                                           ("ohe", ohe_pip, category_columns)],remainder="passthrough") 
preprocessing
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_validate
from sklearn.metrics import recall_score,f1_score,accuracy_score,precision_score
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

models = [
    ("LogisticRegression", LogisticRegression(random_state=42)),
    ("DecisionTreeClassifier", DecisionTreeClassifier(random_state=42)),
    ("KNeighborsClassifier", KNeighborsClassifier()),
    ("GaussianNB", GaussianNB()),
    ("SVC", SVC(random_state=42)),
    ("RandomForestClassifier", RandomForestClassifier(random_state=42))
]

for model in models:
    model_pipeline = Pipeline(steps=[("preprocessing", preprocessing), ("model", model[1])])
    scores = cross_validate(model_pipeline, x, y, cv=5, scoring=["accuracy", "precision", "recall", "f1"], return_train_score=True, n_jobs=-2)
    print(f"Model: {model[0]}")
        # Accuracy Score
    print(f"Train Accuracy: {scores['train_accuracy'].mean().round(4) * 100}%")
    print(f"Test Accuracy: {scores['test_accuracy'].mean().round(4) * 100}%")

    # Precision Score
    print(f"Train Precision: {scores['train_precision'].mean().round(4) * 100}%")
    print(f"Test Precision: {scores['test_precision'].mean().round(4) * 100}%")

    # Recall Score
    print(f"Train Recall: {scores['train_recall'].mean().round(4) * 100}%")
    print(f"Test Recall: {scores['test_recall'].mean().round(4) * 100}%")

    # F1 Score
    print(f"Train F1 Score: {scores['train_f1'].mean().round(4) * 100}%")
    print(f"Test F1 Score: {scores['test_f1'].mean().round(4) * 100}%")

    print('-' * 50)
```

# let's try SMOTE to see if it increase the precentage of percision and recall 

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as imb_pipeline

smote = SMOTE(random_state = 42)

models = [
    ('Logistic Regression', LogisticRegression(random_state= 42)),
    ('KNN', KNeighborsClassifier()),
    ('Gaussian NB', GaussianNB()),
    #('SVC', SVC(random_state= 42)),
    ('Decision Tree', DecisionTreeClassifier(random_state= 42)),
    ('Random Forest', RandomForestClassifier(random_state= 42)),
    ('XGBoost', XGBClassifier(random_state= 42))
]

for model in models:
    
    model_pipeline = imb_pipeline(steps= [('Preprocessing', preprocessing),
                                          ('SMOTE', smote),
                                          ('Model', model[1])])
    
    cv_result = cross_validate(estimator= model_pipeline,
                                X= x,
                                y= y,
                                scoring= ['precision', 'recall', 'f1'],
                                cv= 5,
                                return_train_score= True,
                                n_jobs= -2
                                )
    
    print(model[0])
    print('Train Precision :', cv_result['train_precision'].mean().round(4) * 100)
    print('Test Precision :', cv_result['test_precision'].mean().round(4) * 100)

    print('Train Recall :', cv_result['train_recall'].mean().round(4) * 100)
    print('Test Recall :', cv_result['test_recall'].mean().round(4) * 100)

    
    print('Train f1 :', cv_result['train_f1'].mean().round(4) * 100)
    print('Test f1 :', cv_result['test_f1'].mean().round(4) * 100)
    print('\n')
```

## After using SMOTE i notice that , yes Recall score increased but the percision score become very very low - beside what we are trying to predict if the customer at the end bought from us or not so - if the presicion goes down that's mean that model predict that there is too much customers that didn't buy but the model predict the inverse so the most important evaluation metrics here is F1 and SMOTE didn't affect F1 by the good way so we will neglect it
